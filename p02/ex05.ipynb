```python
# 실행: shift+enter
# 함수명(인자1, 인자2, ...)
# 이스케이프 문자: \n(줄 바꿈), \t
# end는 마지막 인자
print('hello', 'hi', 'ok', end='\t')
print('hello', end='\t')
print('hi')
```

```python
# 데이터(값)? 
# - 숫자(int, float), str, bool, list, tuple, set, dict
# 변수? 그릇
# 연산자: 
# - 산술(+, -, *, /, %, //, **), 대입(=), 
# - 비교(==, !=, >, <, >=, <=)
# 10를 a에 할당(대입)한다.
# 내장 함수(function, 기능): print(), range()
# range(2, 20): 2, 3, 4, ..., 19 -> 정수 
# range(10)
# range(0, 10): 0, 1, 2, ..., 9

# 제어문
# 1. 반복문: for ~ in
""" 
	for 변수 in 반복가능객체:
		문장1
	반복가능객체: range(), 리스트, 튜플, 세트, 딕셔너리
"""
for item in range(2, 20):
	for each in range(2, 20):
		print(item, 'x', each, '=', item * each)
```

```python
# f-string(f-문자열)
for item in range(2, 20):
	for each in range(2, 20):
		# print(f'{item} x {each} = {item * each}')
		# 표현식(변수)은 {}로 묶는다.
		# 표현식? 값이 될 수 있는 식
		print(f'{item} x {each} = {item * each}')
```

```python
# print()
for i in range(1, 10):
    for dan in range(1, 20):
        print(f"{dan:2} × {i} = {dan * i:3}", end="     ")
    print()
```

```python
for dan in range(1, 20):
    print(f"== {dan}단 ==")
    for i in range(1, 10):
        print(f"{dan} x {i} = {dan * i}")
    print()
```
