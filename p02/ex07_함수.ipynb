```python
# isXXX(): True 또는 False
# isinstance(값, 데이터타입): a가 숫자면 True, 아니면 False
# 연산자
# 1. 대입
# 2. 산술: * / -> + -
# 3. 비교
# 4. 논리: not -> and, or

# 함수 정의
def calculator(a, b):
	if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
		return '숫자형만 입력할 수 있습니다.'
	return a + b, a - b, a * b, a / b

# 함수 호출
print(calculator(12, '글자'))
print(calculator(10, 2))
```
