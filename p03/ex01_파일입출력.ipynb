```python
f = open('a.txt', 'r', encoding='utf-8')
print("1:", f.read())
print("2:", f.read())
f.seek(3)
print("3:", f.read())
f.close()
```

```python
f = open('a.txt', 'r', encoding='utf-8')
diary = f.read()
print(diary)
print(diary[:5])
f.close()
```

```python
f = open('abcde.txt', 'w')
f.write('I went to school today.')
print(f)
f.close()
```

```python
# with 문: 자동으로 close() 시킴
with open('abcde.txt', 'w') as f:
    f.write('I went to school today.')
    print(f)
```
