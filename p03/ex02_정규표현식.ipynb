```python
# 정규 표현식(re)
import re
```

```python
# re.match(패턴, 문자열): 문자열 도입에서 패턴 찾기
pattern = 'life'
script = 'life'
print(re.match(pattern, script))
```

```python
# group() 메서드를 사용해 일치하는 내용을 반환
print(re.match(pattern, script).group())
```

```python
print(re.match('life', 'life').group())
```

```python
# 일치하는 패턴이 없으면 에러
# re.match('life', 'animal').group()
```

```python
def refinder(pattern, script):
    if re.match(pattern, script):
        print('Match!')
    else:
        print('Not a match!')
```

```python
script = 'life is so cool'
refinder(pattern, script)
```

```python
# match는 대소문자를 구분한다.
pattern = 'Life'
refinder(pattern, script)
```

```python
# match는 문자열의 처음부터 매칭을 시도한다.
pattern = 'is'
refinder(pattern, script)
```

```python
# re.search(패턴, 문자열): 문자열 전체에서 패턴 찾기
print(re.search(pattern, script).group()) # 문장 중간에 있는 is를 찾는다.
```

```python
print(re.search('life', script).group())
```

```python
print(re.search('cool', script).group())
```

```python
# re.findall(패턴, 찾으려는문자열): 패턴을 찾아서 리스트로 반환
number = 'My number is 511223-1****** and your is 521012-2******'
print(re.findall(r'\d{6}', number)) # 숫자 6개 패턴
```

```python
# 정규표현식의 탐욕 제어하기: 마침표(.)와 물음표(?)
example1 = '저는 91년에 태어났습니다. 97년에는 IMF가 있었습니다. 지금은 2020년입니다.'
# 숫자(\d)로 시작하고, 어떤 문자(.)든 한 번 이상 반복(+)되며, '년'으로 끝나는 문자열 반환
print(re.findall(r'\d.+년', example1))
```

```python
# 탐욕을 멈추는 방법
print(re.findall(r'\d.+?년', example1)) # '년'을 찾으면 멈춤
```

```python
print(re.findall(r'\d+년', example1)) # 숫자 반복 후 '년'으로 끝나는 문자 찾기
```

```python
example = '이동민 교수님은 다음과 같이 설명했습니다(이동민, 2019). 그런데 다른 교수님은 이 문제에 대해서 다는 견해를 가지고 있었습니다(최재영, 2019). 또 다른 견해도 있었습니다(Lion, 2018)'
result = re.findall(r'\((.+?)\)', example)
print(result)
```

```python
# 챗GPT로 정규식 코드 만들기
# 프롬프트:
# 파이썬으로 주어진 텍스트에서 괄호 안의 인용구만 리스트로 출력받는 코드 만들어줘
```

```python
result = re.findall(r'\((.*?)\)', example)
print(result)
```

```python
result = re.findall(r'\(.+\)', example) # .+를 탐욕스럽게 인식
print(result)
```

```python
# ly로 끝나는 단어 추출하기
sentence = 'I have a lovely dog, really. I am not telling a lie'
result = re.findall(r'\w+ly', sentence) # 문자 또는 숫자가 1개 이상 반복하고 ly로 끝나는 문자
print(result)
```

```python
# re.split(패턴, 문자열): 패턴을 기준으로 문자열을 분리
sentence = "I have a lobely dog, really. I am not telling a lie. What a pretty dog! I love this dog."
print(re.split(r'[.?!]', sentence))
```

```python
data = 'a:3; b:4; c:5'
for i in re.split(r'; ', data):
    print(re.split(r":", i))
```

```python
# re.sub(패턴, 대체할문자, 찾을문자열): 찾을 문자열에서 패턴을 찾아 대체할 문자로 교체
print(re.sub('dog', 'cat', sentence)) # sentence에서 dog를 cat으로 교체
```

```python
words = 'I am home now. \n\n\nI am with my cat.\n\n'
print(words)
```

```python
print(re.sub(r'\n', '', words)) # 불필요한 공백 제거
```

```python
# str.replace('찾는문자열', '바꿀문자열')
a_name = '오성실대학생, 김성수대학생'
print(a_name.replace('대학생', '')) # '대학생'을 찾아서 빈 문자열로 대체
```

```python
a_number = '22,333,422'
print(a_number.replace(',', '')) # 쉼표를 없애기
```
