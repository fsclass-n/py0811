```python
# 함수로 만들기
# 함수 정의
# def 함수명(매개변수):
#   문장
#   return 값
def opencsv(filename):
    f = open(filename, 'r', encoding='utf8')
    reader = csv.reader(f)
    output = []
    for i in reader:
        output.append(i)
    return output

# 함수 호출
# 함수명(인자)
result = opencsv('example2.csv')

# 출력
print("37:", result)
```
