```python
# 첫 행 지정하기
new = [['구', '한국인', '외국인', '외국인 비율(%)']]
i = newPop[1]
print("31:", i)

new.append([i[0], i[2], i[3], foreign])
print("34:", new)
```

```python
# 외국인 비율이 3%를 초과할 때만 출력해 보기
for i in newPop:
    try:
        foreign = round(i[3] / (i[2] + i[3]) * 100, 1)
        if foreign > 3:
            print("41:", [i[0], i[2], i[3], foreign])
    except:
        pass
```

```python
new = [['구', '한국인', '외국인', '외국인 비율(%)']]
for i in newPop:
    try:
        foreign = round(i[3] / (i[2] + i[3]) * 100, 1)
        if foreign > 3:
            new.append([i[0], i[2], i[3], foreign])
    except:
        pass
```
