```python
# 부동산 실거래가 살펴보기
# CSV 파일 준비하기: http://rtmobile.molit.go.kr
# CSV형 리스트로 만들고 자료 가공하기
import os, re
import usecsv

os.chdir(r'E:\wi\git\py0811\p04')
apt = usecsv.switch(usecsv.opencsv('apt202605.csv'))
print("7:", apt[:3])
```

```python
# 슬라이싱으로 원하는 자료만 출력하기
print("11:", apt[0])

for i in apt[:6]:
    print("15:", i[1])

for i in apt[:6]:
    print("19:", i[1], i[5], i[9])
```

```python
# 경기도에 120m² 이상 3억 원 이하인 아파트 검색하기
for i in apt:
    try:
        if i[6] >= 120 and i[9] <= 30000 and re.match('경기', i[1]):
            print("26:", i[1], i[5], i[9])
    except:
        pass
```

```python
# 분석 결과를 CSV 파일로 저장하기
new_list = []
for i in apt:
    try:
        if i[6] >= 120 and i[9] <= 30000 and re.match('경기', i[1]):
            new_list.append([i[1], i[5], i[9]])
    except:
        pass
usecsv.writecsv('over120_lower30000.csv', new_list)
```
