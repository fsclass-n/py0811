### 사업성 분석? 
사업을 했을 때 이익이나 손해가 어느 정도 나올지 따져보는 것
### 경제적 타탕성의 개념
- 순현재가치(Net Present Value, NPV) : 할인된 현금 흐름의 값을 모두 더한 값을 나타내는 지표
- 내부수익률(Internal Rate of Return, IRR) : 순현재가치를 0으로 만드는 할인율

### 재무적 타당성의 세부 지표 (p214)
1. 비용(Cost) : 사업을 하는 데 드는 돈
2. 수입(Benefit) : 사업을 통해 얻는 돈
3. 현금 흐름(Cash Flow, CF): 
    - 수입에서 비용을 뺀 값 
    - 즉, CF = B - C
4. 연차(n) : 사업을 하는 기간의 연차, 예를 들어 1년차, 2년차 등 -> 총 사업 기간
5. 할인율(Discount Rate): 
    - 미래의 현금 흐름을 현재 가치로 환산할 때 사용하는 비율
    - 여기서는 사회적 할인율 적용

6. 사회적 할인율(Social Discount Rate)
    - 사회적 관점에서 미래의 가치를 현재 가치로 환산할 때 사용하는 할인율
    - 일반적으로 3% ~ 5% 사이로 설정 -> 2024년 1월 기준 4.5%를 사용

7. 현금 흐름의 현재 가치(Present Value of Cash Flow)
    - 할인율을 적용한 k년차 현금 흐름의 현재가치를 의미한다.
    - CF / (1 + r) ** n, 여기서 r은 할인율, n은 연차

8. 순현재가치(NPV): 
    - 모든 연차의 현금 흐름의 현재 가치를 합한 값, 
    - 즉 NPV = Σ (CF / (1 + r)^n)

```python
# 자본의 현재 가치구하기
import numpy as np

# 할인율 5%
discount = .05     
# 현금 흐름 100억원
cashflow = 100

# 자본의 현재가치 함수
def presentvalue(n):
    return cashflow / (1 + discount) ** n
```

```python
# 1년 후의 현재 가치
print(presentvalue(1))
# 2년 후의 현재 가치
print(presentvalue(2))
```

```python
# 20년 동안 발생한 현재가치
for i in range(20):
    print(presentvalue(i))
```

```python
### 놀이공원 사업의 사업성 분석하기
# 세부 지푯값 구하기

# 놀이공원 사업의 현금 흐름 자료(단위: 억 원) (p217)
# 첫 번째 해(Y0)는 750억 원의 비용만 발생
# 두 번째 해부터 10년 동안 매년 운영 수식 150억 원과 운영비 50억 원이 발생해 총 100억 원의 이익이 발생
```

```python
# 1, 2년 차에 발생한 비용
loss = [-750]
# 3년 차부터 발생한 이익
profit = [100] * 10
print(profit)
```

```python
# 총 10년 동안의 현금 흐름을 리스트로 만들어 cf에 저장
cf = loss + profit
print(cf)
print(len(cf))
```

```python
# cf를 배열로 만들어 cashflow에 저장
cashflow = np.array(cf)
```

```python
# 순현재가치와 내부수익률 함수의 독립(numpy_financial)
import numpy_financial as npf
```

```python
# 순현재가치
# npf.npv(할인율, 현금흐름)
# 할인율은 2023년 예비 타당성 조사에서 사용하는 4.5%를 적용
npv = npf.npv(0.045, cashflow)
print(npv)
```

```python
# 내부수익률? 순현재가치를 0으로 만드는 수익률로, 
# 내부수익률이 사회적 할인율보다 높을 때 공공사업은 수익성이 있다고 말할 수 있다.
# npf.irr(현금흐름)
irr = npf.irr(cashflow)
print(irr)
```
