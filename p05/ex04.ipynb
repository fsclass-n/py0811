```python
import pandas as pd

# 딕셔너리 자료를 data 객체에 저장
data = {'name': ['Mark', 'Jane', 'Chris', 'Ryan'],
        'age': [33, 32, 44, 42],
        'score': [91.3, 83.4, 77.5, 87.7]}

print(data)
```

```python
# DataFrame() 함수로 data 객체를 데이터프레임으로 만들어 df 객체에 저장
df = pd.DataFrame(data)
df
```

```python
# 합계
print(df.sum())
```

```python
# df → DataFrame 객체
# .mean() → 평균 계산
# numeric_only=True → 숫자 데이터만 사용
# print(df.mean()) -> 에러
print(df.mean(numeric_only=True))
```

```python
# 데이터 선택하기
print(df.age)
print(df['age'])
```

```python
# 열 전체 자료 출력하기
print(df['시군구'])
```

```python
# 조건별로 출력하기
# 전용면적이 80을 넘으면 True, 아니면 False
print(df['전용면적'] > 80)
```

```python
# 전용면적이 80을 넘는 데이터프레임만 출력
print(df[df['전용면적'] > 240])
```

```python
# 전용면적이 80m²을 넘는 아파트 단지명 출력
print(df['단지명'][df['전용면적'] > 240])
```

```python
# 전용면적이 130m²가 넘고 거래금액이 3억 원 미만인 아파트 단지명 출력
# 논리 연산자: 
# -> &(그리고 - 모든 조건이 참이면 참)
# -> |(또는 - 하나 이상 참이면 참)
print(df['단지명'][(df['전용면적'] > 130) & (df['거래금액'] < 30000)])
```

```python
# 전용면적이 130m²가 넘거나 거래금액이 3억 원 미만인 아파트 단지명 출력
print(df['단지명'][(df['전용면적'] > 130) | (df['거래금액'] < 30000)])
```

```python
# 원하는 자료만 살펴보기
# df.loc[행선택, 열선택]
# loc는 label-based location indexer

# '단지명'과 '거래금액'만 10개 행출력
# :10 -> 0~10행까지 선택
# ['단지명', '거래금액'] -> 두 컬럼만 선택
print(df.loc[:10, ['단지명', '거래금액']])
```

```python
# 10억 원을 초과하는 가격으로 거래된 아파트
print(df.loc[:, ['단지명', '거래금액']][df['거래금액'] > 100000])
```

```python
# 새로운 값 추가하기
# df['새로운열이름'] = 값

# 전용면적 대비 거래금액을 '단가'라는 이름으로 저장
df['단가'] = df['거래금액'] / df['전용면적']

df.loc[:10, ('거래금액', '전용면적', '단가')]
```

```python
# 데이터 정렬(sort)
# df.sort_values(by='열이름') -> 오름차순
print(df.sort_values(by='거래금액').loc[:, ('거래금액', '시군구')])
```

```python
# df.sort_values(by='열이름', ascending=False) -> 내림차순
print(df.sort_values(by='거래금액', ascending=False).loc[:, ('거래금액', '시군구')])
```

```python
# 4억이 넘는 집만 먼저 추출
# 전용면적 기준으로 오름차순 정렬
# 정렬한 데이터프레임에서 ('거래금액', '전용면적', '시군구') 정보만 추출

print(df[df['거래금액'] > 40000].sort_values(by='전용면적').loc[:, ('거래금액', '전용면적', '시군구')])
```

```python
# 문자열 다루기
# df['검색할열'].str.contains('찾는문자열')

# '시군구'에서 '강릉'을 포함한 거래는 True 반환, 없다면 False
print(df['시군구'].str.contains('강릉'))
```

```python
print(df[df['시군구'].str.contains('강릉')])
```
