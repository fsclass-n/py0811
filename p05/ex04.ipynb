```python
import pandas as pd

# 딕셔너리 자료를 data 객체에 저장
data = {'name': ['Mark', 'Jane', 'Chris', 'Ryan'],
        'age': [33, 32, 44, 42],
        'score': [91.3, 83.4, 77.5, 87.7]}

print(data)
```

```python
# DataFrame() 함수로 data 객체를 데이터프레임으로 만들어 df 객체에 저장
df = pd.DataFrame(data)
df
```

```python
# 합계
print(df.sum())
```

```python
# df → DataFrame 객체
# .mean() → 평균 계산
# numeric_only=True → 숫자 데이터만 사용
# print(df.mean()) -> 에러
print(df.mean(numeric_only=True))
```

```python
# 데이터 선택하기
print(df.age)
print(df['age'])
```
