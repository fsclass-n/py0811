```python
df2 = pd.read_csv('survey.csv')
print(df2.head())
```

```python
# 기초 통계량 구하기
# DataFrame의 기본 통계 요약(summary statistics)을 자동으로 보여주는 메서드
print(df2.describe())
```

```python
# 수입의 기초통계량
print(df2['income'].describe())
```

```python
# 수입의 합계
print(df2['income'].sum())
```

```python
# 수입의 중앙값
print(df2['income'].median())
```

```python
# 기초 통계량 추가 분석하기
# 빈도(frequency) 분석하기: value_counts()
# df['변수'].value_counts()

# 남성과 여성의 빈도
print(df2['sex'].value_counts())

# 결과: 남성 14명, 여성 9명이 설문에 응했다.
```

```python
# 두 집단의 평균 구하기: groupby()
# df.groupby(그룹을나누는변수).연산
print(df2.groupby('sex').mean())
```

```python
print(df2.groupby('sex')['income'].mean())
# 결과: 여성의 평균 수입은 4,333만 원, 남성의 평균 수입은 4,285만 원으로 나왔다.
```
