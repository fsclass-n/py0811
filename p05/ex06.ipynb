```python
# 1. import 모듈
import os
# 2. import 모듈 as 별칭
import pandas as pd

# 패키지 > 모듈(.py) > 클래스,함수,변수
# pip install scipy
# SciPy 라는 파이썬 과학 계산 라이브러리
# 3. from 패키지(모듈) import 모듈(함수)
from scipy import stats
```

```python
df2 = pd.read_csv('survey.csv')
male = df2.income[df2.sex == 'm']
female = df2.income[df2.sex == 'f']

print("13:", stats.ttest_ind(male, female))

# 13: TtestResult(statistic=np.float64(-0.10650308143428425), pvalue=np.float64(0.916194078116337), df=np.float64(21.0))
```

```python
# pvalue는 유의확률(p-value)을 의미한다.
# 유의확률값이 작을수록 유의한 차이가 있다고 해석한다.
# 일반적으로 95% 또는 99%를 유의한 확률의 기준으로 삼기 때문에 
# 유의확률값이 0.05 미만이거나 0.01 미만이면 유의한 차이가 나타난다고 말할 수 있다.

# 결과: 현재 pvalue는 0.916으로 1에 가까울 정도로 높다.
# 즉, 남성과 여성의 수입 평균을 비교한 t검정 결과에 유의한 차이가 있다고 보기 어렵다.
# 유의확률, 즉 pvalue가 0.05보다 작은 경우에만 유의하다고 판정
```

```python
ttest_result = stats.ttest_ind(male, female)
if ttest_result[1] > .05:
    print('p-value는 %f로 95%% 수준에서 유의하지 않음' % ttest_result[1])
else:
    print('p-value는 %f로 95%% 수준에서 유의함' % ttest_result[1])
```

```python
### 상관관계(corelation) 분석? 
# 두 변수가 어떤 관계인지를 알아보는 방법이다.

#### 상관관계를 분석하는 방법
# - 피어슨(Pearson)
# - 스피어만(Spearman)

#### 상관계수(r)? 
# - 상관관계 분석 결과로 도출되는 값
# - 상관계수(r)의 범위는 -1부터 1까지이다.
# - 스피어만 상관계수는 변수가 순위 척도(ordinal scale)일 때 사용한다.
#     - 예) 어떤 고등학교의 3학년 모의고사 성적을 과목별로 등수를 매겼을 때 국어 영역의 등수와 수리 영역의 등수가 어떤 상관관계가 있는지 스피어만의 상관관계로 알아볼 수 있다.
# - 피어슨의 상관계수는 순위가 아니라 연속형 자료(continuous data)의 상관관계를 다룬다.
# - 연속형 자료? 점수, 키, 성적 등에서 연속된 숫자로 이루어진 자료
```

```python
# 두 변수의 상관관계 분석하기
# 피어슨 상관관계 분석 결과 출력
# df2[['변수1, '변수2']].corr()

# 수입(income)과 직업만족도(jobSatisfaction)의 상관관계
print("52:", df2[['income', 'jobSatisfaction']].corr())
```

```python
# 스피어만 상관관계 분석 결과 출력
# df2[['변수1', '변수2'].corr(method="spearman")]

# 비모수 기법인 스피어만의 순서형 상관관계로 수입, 직업만족도, 스트레스의 상관관계
print(df2[['income', 'jobSatisfaction', 'stress']].corr(method="spearman"))
```

```python
# 수입과 스트레스의 상관관계
print(df2[['income', 'stress']].corr())
```

```python
# 결과: income과 stress의 상관계수는 -0.13792
# 음의 상관관계는 두 변수의 변화 방향이 반대
# 즉, 스트레스와 수입이 반비례 관계
# 단, 가공한 자료이므로 특별한 의미는 없다.
```

```python
# 회귀 분석(regression)
# statsmodels 패키지로 회귀 분석하기
# pip install statsmodels
# import 패키지명.하위패키지명.모듈(.py) as 별칭
import statsmodels.formula.api as smf

# ols(formula='종속변수~독립변수', data=데이터프레임)
# 종속변수로 직업만족도, 독립변수로 영어성적을 입력
model = smf.ols(formula='jobSatisfaction~English', data=df2)
result = model.fit()

# 회귀 분석 결과 확인
print(result.summary())
```

```python
# R-squared: 설명력
# Prob(F-statistic): 유의확률
# coef(coefficient): 계숫값
# Prob(Omnibus): 총괄 검정 유의확률
```

```python
# 다중 회귀 분석
# smf.ols(formula='종속변수~독립변수1 + ... + 독립변수n', data=데이터프레임)
model2 = smf.ols(formula='jobSatisfaction~English + stress + income', data=df2)
result = model2.fit()
print(result.summary())
```
