```python
# 맷플롯립으로 데이터 시각화하기
# 그래프 만들고 출력하기

# from 패키지 import 모듈 as 별칭
from matplotlib import pyplot as plt
# import 패키지.모듈 as 별칭
# import matplotlib.pyplot as plt
```

```python
# 데이터
X = [1, 4, 9, 16, 25, 36, 49, 64]

plt.plot(X)
plt.show()
```

```python
# plt.plot(그래프자료, 모양+색)
plt.plot(X, color='b')
plt.show()
```

```python
print(plt.plot(X, 'or'))
plt.show()
```

```python
# 리스트 값 출력
print(X)
```

```python
y = [i for i in range(1, 9)]
print(y)
```

```python
# x축 이름 지정하기
plt.xlabel('x_name')
# y축 이름 지정하기
plt.ylabel('y_name')
# 그래프 제목
plt.title('matplotlib sample')
plt.plot(X, y)
plt.show()
```
